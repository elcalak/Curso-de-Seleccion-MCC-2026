Algunos dispositivos solo cuentan con unos pocos botones, muchos menos que el número de letras del alfabeto. En estos dispositivos, pueden ser necesarias varias pulsaciones para escribir un solo carácter. Cuatro botones de flecha controlan el movimiento del cursor y, cuando este se sitúa sobre la tecla adecuada, al pulsar el quinto botón se selecciona el carácter correspondiente y se añade al final del texto. Para finalizar el texto, el usuario debe desplazarse hasta la tecla «Intro» y seleccionarla. Esto proporciona a los usuarios un conjunto arbitrario de caracteres y les permite escribir texto de cualquier longitud con solo cinco botones físicos.

En este problema, se te proporciona una disposición de teclado virtual y tu tarea consiste en determinar el número mínimo de pulsaciones necesarias para escribir un texto dado, donde pulsar cualquiera de los cinco botones físicos constituye una pulsación. El cursor comienza en la esquina superior izquierda del teclado y se desplaza en las cuatro direcciones cardinales, de tal manera que siempre salta al siguiente cuadrado unitario en esa dirección que pertenezca a una tecla diferente. La primera línea de la entrada contiene dos números enteros r y c (1 ≤ r, c ≤ 50), que indican el número de filas y columnas de la cuadrícula del teclado virtual. El teclado virtual se especifica en las siguientes r líneas, cada una de las cuales contiene c caracteres. Los valores posibles de estos caracteres son letras mayúsculas, dígitos, un guión y un asterisco (que representa la tecla Intro). Solo hay una tecla correspondiente a cada carácter dado. Cada tecla está formada por una o más casillas de la cuadrícula, que siempre formarán una región contigua. La última línea de la entrada contiene el texto que hay que escribir. Este texto es una cadena no vacía de como máximo 10 000 de los caracteres disponibles, excepto el asterisco.

# Output

Muestra el número mínimo de pulsaciones necesarias para escribir el texto completo, incluida la tecla Intro al final. Se garantiza que el texto se puede escribir.

# Problema L
## Weather Report
### Time limit: 2 Segundos

Has sido contratado por la Asociación para la Medición Climatológica, una organización científica interesada en el seguimiento de las tendencias meteorológicas globales a lo largo de un periodo prolongado. Por supuesto,  no es una tarea fácil. Han instalado numerosos dispositivos pequeños por todo el mundo, diseñados para realizar mediciones periódicas de las condiciones meteorológicas locales. Se trata de dispositivos económicos Con capacidades algo limitadas. Cada día observan cuál de los cuatro tipos de tiempo estándar se ha producido: soleado, nublado, lluvioso o ranas. Una vez realizadas n de estas observaciones, los resultados se envían al servidor principal para su análisis. Sin embargo, el enorme número de dispositivos ha provocado una sobrecarga del ancho de banda de comunicación disponible. La Asociación necesita tu ayuda para idear un método que permita comprimir estos informes en menos bits. Para la ubicación de un dispositivo concreto, puedes suponer que el tiempo de cada día es un evento aleatorio independiente y se te proporcionan las probabilidades previstas de los cuatro tipos de tiempo posibles. Cada uno de los 4n informes meteorológicos posibles para un dispositivo debe codificarse como una secuencia única de bits, de tal forma que ninguna secuencia sea un prefijo de ninguna otra secuencia (una propiedad importante, ya que, de lo contrario, el servidor no sabría cuándo termina cada secuencia). El objetivo es utilizar una codificación que minimice el número esperado de bits transmitidos.

# Input

La primera línea de entrada contiene un número entero 1 ≤ n ≤ 20, que indica el número de observaciones que se incluyen en cada informe. La segunda línea contiene cuatro números de coma flotante positivos, psunny, pcloudy, prainy y pfrogs, que representan las respectivas probabilidades meteorológicas. Estas probabilidades tienen como máximo 6 dígitos después del punto decimal y su suma es 1.

# Output

Indica el número mínimo esperado de bits en la codificación de un informe, con un error absoluto o relativo de como máximo 10.

# Problema M
## Window Manager
### Time limit: 2 Segundos

En los últimos años se ha producido una revolución en la tecnología de las interfaces de usuario. Durante muchos años, los teclados y los ratones fueron las herramientas utilizadas para interactuar con los ordenadores. Sin embargo, con la llegada de los teléfonos inteligentes y las tabletas, cada vez es más habitual que los usuarios utilicen sus ordenadores tocando y deslizando los dedos por la pantalla. Como es lógico, esto ha dado lugar a nuevos paradigmas en el diseño de interfaces de usuario. Un principio importante es que los objetos que aparecen en la pantalla obedezcan a leyes «físicas». En este problema, verás un ejemplo de ello. Te han contratado para crear un simulador del gestor de ventanas que se utilizará en la próxima generación de teléfonos inteligentes de Advanced Cellular Manufacturers (ACM). Cada teléfono que fabriquen tendrá una pantalla rectangular en la que se mostrarán completamente cero o más ventanas rectangulares. Es decir, ninguna ventana sobrepasa los límites de la pantalla ni se superpone a ninguna otra ventana. El simulador debe admitir los siguientes comandos.

• OPEN x y w h — abre una nueva ventana con coordenadas de la esquina superior izquierda (x, y), anchura w píxeles y altura h píxeles.

• CLOSE x y — cierra una ventana abierta que incluya el píxel en (x, y). Esto permite al usuario tocar en cualquier punto de una ventana para cerrarla.

• RESIZE x y w h — establece las dimensiones de la ventana que incluye el píxel en (x, y) en un ancho de w y una altura de h. La esquina superior izquierda de la ventana no se mueve.

• MOVE x y dx dy — desplaza la ventana que incluye el píxel situado en (x, y). El desplazamiento es de dx píxeles en dirección horizontal o de dy píxeles en dirección vertical. Como máximo, uno de los valores dx y dy será distinto de cero.

Los comandos OPEN y RESIZE solo se ejecutan con éxito si la ventana resultante no se superpone a ninguna otraventana y no se extiende más allá de los límites de la pantalla. El comando MOVE desplazará la ventana tantos píxeles como sea posible de los solicitados. Por ejemplo, si dx es 30 pero la ventana solo puede desplazarse 15 píxeles hacia la derecha, entonces se desplazará 15 píxeles. ACM se enorgullece especialmente del comando MOVE. Una ventana que se está moviendo podría «chocar» con otra ventana. En este caso, la primera ventana empujará a la segunda en la misma dirección hasta donde sea necesario, exactamente como si las ventanas fueran objetos físicos. Este comportamiento puede producirse en cascada: una ventana en movimiento podría encontrarse con otras ventanas que también serían empujadas según sea necesario. La figura M.1 muestra un ejemplo con tres ventanas, en el que la ventana A se desplaza hacia la derecha, empujando a las otras dos con ella.

# Input

La primera línea de entrada contiene dos números enteros positivos, xmax e ymax, que representan las dimensiones horizontal y vertical de la pantalla, medidas en píxeles. Cada uno de ellos es, como máximo, 109 (la ACM tiene previsto fabricar pantallas con una resolución muy alta). El píxel superior izquierdo de la pantalla tiene coordenadas (0, 0). Cada una de las siguientes líneas contiene un comando tal y como se ha descrito anteriormente. Uno o más espacios separan el nombre del comando y los parámetros entre sí. Los parámetros del comando son números enteros que satisfacen estas condiciones: 0 ≤ x < xmax, 0 ≤ y < ymax, 1 ≤ w, h ≤ 10⁹, y |dx|, |dy| ≤ 10⁹. Habrá como máximo 256 comandos.

# Output:

La salida debe seguir el formato que se muestra en el ejemplo de salida que figura a continuación.

Simula los comandos en el orden en que aparecen en la entrada. Si se detecta algún error durante la simulación de un
comando, muestra el número del comando, el nombre del comando y el primer mensaje adecuado de la siguiente lista, e ignora los resultados de la simulación de ese comando (salvo que se indique lo contrario).

• no hay ventana en la posición indicada —para los comandos CLOSE, RESIZE y MOVE— sino hay ninguna ventana que incluya el píxel en la posición especificada.

• la ventana no cabe —para los comandos OPEN y RESIZE— si la ventana resultante se superpondría a otra ventana o se extendería más allá de los límites de la pantalla.

Los valores d y d′ son el número absoluto de píxeles solicitados y desplazados, respectivamente. En este caso, la ventana sigue desplazándose, pero solo una distancia menor.

Una vez simulados todos los comandos y mostrados los mensajes de error, indica el número de ventanas que siguen abiertas. A continuación, para cada ventana abierta, en el mismo orden en que se abrieron, muestra las coordenadas de la esquina superior izquierda (x, y), la anchura y la altura.
