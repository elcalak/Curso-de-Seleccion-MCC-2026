# Problema A
## Amalgamated Artichokes
### Time limit: 5 Segundos

Fátima Cynara es analista en Amalgamated Artichokes (AA). Al igual que cualquier empresa, AA ha pasado por momentos muy buenos y otros no tan buenos. Fátima realiza análisis de tendencias de las cotizaciones bursátiles de AA y quiere determinar la mayor caída de las cotizaciones a lo largo de distintos periodos de tiempo. Por ejemplo, si durante un periodo de tiempo los precios de las acciones fueran 19, 12, 13, 11, 20 y 14, la mayor caída sería de 8 entre el primer y el cuarto precio. Si el último precio hubiera sido 10 en lugar de 14, la mayor caída habría sido de 10 entre los dos últimos precios.

Fátima ha realizado algunos análisis previos y ha descubierto que el precio de las acciones durante cualquier periodo de tiempo puede modelarse con bastante precisión mediante la siguiente ecuación:

price(k) = p · (sin(a · k + b) + cos(c · k + d) + 2)

donde p, a, b, c y d son constantes. A Fátima le gustaría que escribieras un programa para determinar la mayor caída de precios en una secuencia dada de precios. La figura A.1 ilustra la función de precios para la entrada de ejemplo 1. Debes tener en cuenta únicamente los precios correspondientes a valores enteros de k

# Input

La entrada consta de una sola línea que contiene 6 números enteros: p (1 ≤ p ≤ 1 000), a, b, c, d (0 ≤ a, b, c, d ≤
1 000) y n (1 ≤ n ≤ 10⁶). Los primeros 5 números enteros se han descrito anteriormente. La secuencia de cotizaciones bursátiles a
tener en cuenta es price(1), price(2), …, price(n).

# Output

Muestra la caída máxima de los precios de las acciones. Si no se produce ninguna caída, muestra el número 0. Tu
resultado debe tener un error absoluto o relativo de, como máximo, 10−6.

# Problema C

# Catering

## Time limit: 4 segundos

Paul es propietario de una empresa de catering y el negocio va viento en popa. La empresa cuenta con k equipos de catering, cada uno de ellos a cargo de un conjunto de equipo de catering.
Cada semana, la empresa acepta n solicitudes de catering para diversos eventos. Por cada solicitud, envían un equipo de catering con su equipo al lugar del evento. El equipo entrega la comida, monta el equipo e instruye al anfitrión sobre cómo utilizarlo y servir la comida. Tras el evento, el anfitrión se encarga de devolver el equipo a la empresa de Paul. Desgraciadamente, algunas semanas el número de equipos de catering es inferior al número de solicitudes, por lo que es posible que algunos equipos tengan que utilizarse para más de un evento. En estos casos, la empresa no puede esperar a que el anfitrión devuelva el equipo y debe mantener al equipo en el lugar para trasladar el equipo a otra ubicación. La empresa dispone de una estimación precisa del coste que supone trasladar un conjunto de equipos desde cualquier ubicación a cualquier otra. Teniendo en cuenta estos costes, Paul quiere elaborar un «Mapa de catering anticipado» para atender las solicitudes minimizando al mismo tiempo el coste total de traslado del equipo (incluido el coste del primer traslado), aunque eso signifique no utilizar todos los equipos disponibles. Paul necesita tu ayuda para escribir un programa que realice esta tarea. Las solicitudes están ordenadas en orden ascendente según la hora de sus eventos y se eligen de tal manera que, para cualquier i < j, haya tiempo suficiente para transportar el equipo utilizado en la i-ésima solicitud a la ubicación de la j-ésima solicitud.

# Input

La primera línea de la entrada contiene dos números enteros n (1 ≤ n ≤ 100) y k (1 ≤ k ≤ 100), que representan el número de solicitudes y el número de equipos de catering, respectivamente. A continuación hay n líneas, en las que la i-ésima línea contiene n − i + 1 números enteros comprendidos entre 0 y 1 000 000, ambos inclusive. El j-ésimo número de la i-ésima línea es el coste de trasladar un conjunto de equipos desde la ubicación i a la ubicación i + j. La empresa se encuentra en la ubicación 1 y las n solicitudes se encuentran en las ubicaciones 2 a n + 1.

# Output

Muestra el coste mínimo de traslado para atender todas las solicitudes. (Esta cantidad no incluye el coste de devolver el equipo a la empresa de catering.)

# Problema D

# Cuttin Cheese

## Time limit: 3 segundos

Seguro que todos habéis oído hablar de la International Cheese Processing Company. Su máquina para cortar un trozo dequeso en lonchas de exactamente el mismo grosor es todo un clásico.Recientemente han fabricado una máquina capaz de cortar un queso esférico (como el Edam) en lonchas; no, no todas del mismo grosor, ¡pero sí todas del mismo peso! Sin embargo, se avecinan nuevos retos: cortar queso suizo. El queso suizo, como el Emmental, tiene agujeros, y estos
pueden ser de diferentes tamaños.Una loncha con agujeros contiene menos queso y pesa menos que una loncha sin agujeros. Así pues, este es el reto: cortar un queso con agujeros en lonchas de igual peso. Mediante técnicas inteligentes de sonar (las mismas que se utilizan para explorar fetos y yacimientos petrolíferos), es posible localizar los agujeros del queso con una precisión de micrómetros. Para este problema, puedes suponer que los agujeros son esferas perfectas. Cada bloque sin cortar tiene unas dimensiones de 100 × 100 × 100, donde cada dimensión se mide en milímetros. Tu tarea consiste en cortarlo en s lonchas de igual peso. Las lonchas tendrán 100 mm de ancho y 100 mm de alto, y tu trabajo consiste en determinar el grosor de cada loncha.

# Input

La primera línea de la entrada contiene dos números enteros n y s, donde 0 ≤ n ≤ 10 000 es el número de agujeros que tiene el queso, y 1 ≤ s ≤ 100 es el número de lonchas que hay que cortar. Las siguientes n líneas contienen cada una cuatro números enteros positivos r, x, y y z que describen un agujero, donde r es el radio y x, y y z son las coordenadas del centro, todas en micrómetros. El bloque de queso ocupa los puntos (x, y, z) donde 0 ≤ x, y, z ≤ 100 000, excepto los puntos que forman parte de algún agujero. Los cortes se realizan perpendicularmente al eje z. Puedes suponer que los agujeros no se solapan, pero sí pueden tocarse, y que los agujeros están completamente contenidos en el queso, aunque pueden tocar su contorno.

# Output

Muestra los espesores de las s lonchas en milímetros, empezando por el extremo del queso donde z = 0. Tu salida debe tener un error absoluto o relativo de, como máximo, 10−6.

# Problema E

# Evolution in Parallel

## Time limit: 2 segundos

Estamos en el año 2178 y se ha descubierto vida extraterrestre en un planeta lejano. Parece que solo hay una especie en el planeta y que no se reproduce como lo hacen los animales de la Tierra. Y lo que es aún más sorprendente, ¡la composición genética de todos y cada uno de los organismos es idéntica! La composición genética de cada organismo es una única secuencia de nucleótidos. Los nucleótidos son de tres tipos, denotados por «A» (adenina), «C» (citosina) y «M» (muamina). Según una hipótesis, la evolución en este planeta se produce cuando se inserta un nuevo nucleótido en algún punto de la secuencia genética de un organismo existente. Si este cambio resulta ventajoso desde el punto de vista evolutivo, los organismos con la nueva secuencia sustituyen rápidamente a los que tienen la secuencia antigua. Inicialmente se pensaba que las especies actuales habían evolucionado de esta manera a partir de un único organismo muy simple con una secuencia genética de un solo nucleótido, mediante mutaciones como las descritas anteriormente. Sin embargo, las pruebas fósiles sugieren que quizá no haya sido así. En estos momentos, el equipo de investigación con el que trabajas está tratando de validar el concepto de «evolución paralela»: que en realidad podría haber habido dos vías evolutivas que evolucionaron de la forma descrita anteriormente y que, finalmente, ambas vías dieron lugar a la única especie presente hoy en día en el planeta. Tu tarea consiste en verificar si la hipótesis de la evolución paralela es coherente con el material genético encontrado en las muestras fósiles recogidas por tu equipo.

# Input

La entrada comienza con un número n (1 ≤ n ≤ 4 000) que indica el número de secuencias de nucleótidos encontradas en los fósiles. La segunda línea describe la secuencia de nucleótidos de la especie que vive actualmente en elplaneta. Cada una de las siguientes n líneas describe una secuencia de nucleótidos encontrada en los fósiles. Cada secuencia de nucleótidos consiste en una cadena de al menos una letra, pero no más de 4 000. Las cadenas solo contienen letras mayúsculas A, C y M. Todas las secuencias de nucleótidos, incluida la de la especie actualmente viva, son distintas.

# Output

Muestra un ejemplo de cómo las secuencias de nucleótidos del registro fósil participan en dos vías evolutivas. El ejemplo debe comenzar con una línea que contenga dos números enteros s1 y s2, que representan el número de secuencias de nucleótidos del registro fósil que participan en la primera vía y en la segunda vía, respectivamente. A continuación, deben aparecer s1 líneas que contengan las secuencias atribuidas a la primera vía, en orden cronológico (empezando por la más antigua), y después s2 líneas que contengan las secuencias atribuidas a la segunda vía, también en orden cronológico. Si hay varios ejemplos, muestra cualquiera de ellos. Si es posible que una secuencia pueda aparecer en la historia genética de ambas especies, tu ejemplo debe asignarla a exactamente una de las vías evolutivas. Si es imposible que todo el material fósil proceda de dos vías evolutivas, muestra la palabra «imposible»

# Problema F

# Keyboarding

## Time limit: 4 segundos

¿Cuántas pulsaciones se necesitan para escribir un mensaje de texto? Podrías pensar que es igual al númerode caracteres del texto, pero esto solo es cierto si una pulsación genera un carácter. En los dispositivos de tamaño de bolsillo, las posibilidades para escribir texto suelen ser limitadas. Algunos dispositivos solo cuentan con unos pocos botones, muchos menos que el número de letras del alfabeto. En estos dispositivos, pueden ser necesarias varias pulsaciones para escribir un solo carácter. Un mecanismo para hacer frente a estas limitaciones es un teclado virtual que se muestra en la pantalla, con un cursor que se puede mover de tecla en tecla para seleccionar caracteres. Cuatro botones de flecha controlan el movimiento del cursor y, cuando este se sitúa sobre la tecla adecuada, al pulsar el quinto botón se selecciona el carácter correspondiente y se añade al final del texto. Para finalizar el texto, el usuario debe desplazarse hasta la tecla «Intro» y seleccionarla. Esto proporciona a los usuarios un conjunto arbitrario de caracteres y les permite escribir texto de cualquier longitud con solo cinco botones físicos.

En este problema, se te proporciona una disposición de teclado virtual y tu tarea consiste en determinar el número mínimo de pulsaciones necesarias para escribir un texto dado, donde pulsar cualquiera de los cinco botones físicos constituye una pulsación. Las teclas están dispuestas en una cuadrícula rectangular, de tal forma que cada tecla virtual ocupa uno o más cuadrados unitarios contiguos de la cuadrícula. El cursor comienza en la esquina superior izquierda del teclado y se desplaza en las cuatro direcciones cardinales, de tal manera que siempre salta al siguiente cuadrado unitario en esa dirección que pertenezca a una tecla diferente. Si no existe tal cuadrado unitario, el cursor no se mueve.

# Input

La primera línea de la entrada contiene dos números enteros r y c (1 ≤ r, c ≤ 50), que indican el número de filas y columnas de la cuadrícula del teclado virtual. El teclado virtual se especifica en las siguientes r líneas, cada una de las cuales contiene c caracteres. Los valores posibles de estos caracteres son letras mayúsculas, dígitos, un guión y un asterisco (que representa la tecla Intro). Solo hay una tecla correspondiente a cada carácter dado. Cada tecla está formada por una o más casillas de la cuadrícula, que siempre formarán una región contigua. La última línea de la entrada contiene el texto que hay que escribir. Este texto es una cadena no vacía de como máximo 10 000 de los caracteres disponibles, excepto el asterisco.

# Output

Muestra el número mínimo de pulsaciones necesarias para escribir el texto completo, incluida la tecla Intro al final. Se garantiza que el texto se puede escribir.
